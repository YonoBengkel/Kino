```python
import os

# Jika dijalankan di Google Colab, ambil API Key dari tab Secrets
try:
    from google.colab import userdata
    os.environ['GROQ_API_KEY'] = userdata.get('GROQ_API_KEY')
except ImportError:
    pass # Gunakan konfigurasi .env biasa jika berjalan di lokal

import core

print("=========================================================")
print(" 🎬 KINO - KURATOR RUTE SINEMA ")
print("=========================================================")
print("Halo! Sebutkan film yang kamu suka, dan aku akan memberikan")
print("rute tontonan selanjutnya berdasarkan elemen yang kamu nikmati.\n")
print("💡 Ketik 'exit' untuk keluar.")
print("=========================================================\n")

try:
    klien = core.buat_klien()
    riwayat = core.riwayat_baru()
    
    while True:
        masukan = input("Kamu: ").strip()
        if not masukan:
            continue
            
        if masukan.lower() == 'exit':
            print("\nKino: Selamat menonton! Sampai jumpa lagi.")
            break
            
        print("\nKino: ", end="")
        try:
            # Memanggil fungsi tanya dari core.py yang sudah memiliki fitur streaming
            core.tanya(klien, riwayat, masukan, model='llama-3.3-70b-versatile', suhu=0.7, max_tokens=1024)
        except Exception as e:
            print(f"\n[Error]: {e}")
        print("\n" + "-"*50 + "\n")

except Exception as e:
    print(f"Gagal memulai chatbot: {e}")
    print("Pastikan GROQ_API_KEY sudah terisi dengan benar.")
```
